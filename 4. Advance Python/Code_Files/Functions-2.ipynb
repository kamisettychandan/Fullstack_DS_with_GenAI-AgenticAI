# Variable Length

```python
def sum(a,b):
    c = a + b
    return c
sum(2,3)
```

```python
def sum(a,b):
    c = a + b
    return c
sum(5,6,7,8)   # we define 2 arguments but calling 4 values
```

```python
def sum(a, *b):
    c = a + b
    return c
sum(5,6,7,8)
```

```python
def sum(a, *b):  # 1st argument is fixed but for 2nd argument
    print(type(a))
    print(type(b))
sum(5,6,7,8)
```

```python
def sum(a, *b): # 1st argument is fixed & we fetch each value from the tuple & we can add them. 
    c = a
    for i in b:
        c = c + i
    print(c)
sum(5,6,7,8,9,10,11,100)
```

# kwargs

```python
def person(name, *data):
    print(name)
    print(data)
person('Vihari', 3, 'Nandan', 903050)
```

```python
def person(name,*data):
    print('name')
    print(data)

person('Vihari', age = 3, home_place ='southcity', mob =903050)
```

```python
def person(name, **data):
    print('name')
    print(data)

person('Vihari', age = 3, home_place ='southcity', mob =903050)
# we got error as keyword argument thats why we add another *
```
